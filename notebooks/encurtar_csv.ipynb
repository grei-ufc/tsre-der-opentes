# Código simples desenvolvido com IA para diminuir o tamanho dos datasets. 
Inicialmente deixados com a duração de 15 dias, entretanto os dados do primeiro dia são inconsistentes, portanto foram usados a partir do segundo dia. Logo, te-se 14 dias disponíveis.

```python
import csv
import os
import pandas as pd
from pathlib import Path

BASE_DIR = Path("..").resolve()
pasta = BASE_DIR/'data'/'InfoPV'/'solar_station'
n_de_dias = 15      # Mude aqui a quantidade desejada
linhas_manter = ((96*n_de_dias)+1)  

# Percorre os números de 1 a 51
for numero in range(1, 52):
    # Formata o nome do arquivo (PS_001.csv, PS_002.csv, ...)
    nome_arquivo = f"PS_{numero:03d}.csv"
    caminho_completo = os.path.join(pasta, nome_arquivo)
    
    # Verifica se o arquivo existe
    if os.path.exists(caminho_completo):
        # Lê e escreve no mesmo arquivo
        with open(caminho_completo, 'r', encoding='utf-8') as entrada:
            linhas = list(csv.reader(entrada))
            linhas = linhas[:linhas_manter]
        
        with open(caminho_completo, 'w', newline='', encoding='utf-8') as saida:
            escritor = csv.writer(saida)
            escritor.writerows(linhas)
        
        print(f"✓ Arquivo {nome_arquivo} reduzido para {linhas_manter} linhas!")
    else:
        print(f"✗ Arquivo {nome_arquivo} não encontrado")

print("\nProcessamento concluído!")
```
